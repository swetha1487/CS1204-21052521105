_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('data/diabetes.csv')

plt.figure(figsize=(8, 5))
data['Glucose'].plot(kind='hist', bins=10, color='red', edgecolor='black')  # bins = number of bars
plt.title('Histogram of Glucose')
plt.xlabel('Glucose'); plt.ylabel('Frequency')
plt.grid(True); plt.show()

plt.figure(figsize=(8, 5))
plt.hist(data['Age'], bins=15, color='teal', edgecolor='black')  # histogram using plt.hist
plt.title('Histogram of Age')
plt.xlabel('Age'); plt.ylabel('Frequency')
plt.grid(True); plt.show()
```
