**Step 1:** Start the program.

**Step 2:** Import pandas, numpy, matplotlib.pyplot and Axes3D from mpl_toolkits.mplot3d.

**Step 3:** Enable inline plotting using %matplotlib inline.

**Step 4:** Read the diabetes.csv file into a DataFrame.

**Step 5:** Create a figure using plt.figure() with a 3-D subplot (projection='3d').

**Step 6:** Extract Age, Glucose and Outcome columns as x, y and z arrays.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

data = pd.read_csv('data/diabetes.csv')

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')           # 3-D axes

x = data['Age'].values
y = data['Glucose'].values
z = data['Outcome'].values

ax.set_xlabel("Age (Year)")
ax.set_ylabel("Glucose (Reading)")
ax.set_zlabel("Outcome (0 or 1)")
ax.set_title("3D Scatter - Diabetes Dataset")

ax.scatter(x, y, z, c='blue', marker='^', alpha=0.6)  # 3-D scatter; alpha = transparency
plt.show()
```
