## AIM

To apply Feature Engineering techniques - Feature Scaling (Standard, MinMax) and Categorical Encoding (Label, One-Hot) - on a sample dataset.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, numpy, matplotlib and preprocessing modules from sklearn.

**Step 3:** Create a sample DataFrame with Age, Salary, Experience, Education and City columns.

**Step 4:** Apply StandardScaler on the numeric columns and display the scaled data.

**Step 5:** Apply MinMaxScaler on the numeric columns and display the normalized data.

**Step 6:** Apply LabelEncoder on the ordinal column Education.

**Step 7:** Apply one-hot encoding on the nominal column City using pd.get_dummies().

**Step 8:** Combine the engineered features into a final DataFrame.

**Step 9:** Plot histograms comparing original Salary vs StandardScaled vs MinMaxScaled.

**Step 10:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

data = {
    'Age'       : [22, 25, 47, 52, 46, 56, 35, 28, 60, 40],
    'Salary'    : [25000, 30000, 85000, 95000, 80000, 100000, 55000, 32000, 110000, 60000],
    'Experience': [1, 2, 15, 20, 14, 25, 10, 3, 30, 12],
    'Education' : ['School','UG','PG','PhD','PG','PhD','UG','School','PhD','PG'],
    'City'      : ['Chennai','Mumbai','Delhi','Chennai','Bangalore',
                   'Mumbai','Chennai','Delhi','Bangalore','Mumbai']
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)

numeric_cols = ['Age', 'Salary', 'Experience']

df_std = pd.DataFrame(StandardScaler().fit_transform(df[numeric_cols]),  # mean 0, std 1
                      columns=[c + '_std' for c in numeric_cols])
print("StandardScaled Data (mean=0, std=1):")
print(df_std.round(2))

df_mm = pd.DataFrame(MinMaxScaler().fit_transform(df[numeric_cols]),  # range 0 to 1
                     columns=[c + '_mm' for c in numeric_cols])
print("MinMaxScaled Data (range 0 to 1):")
print(df_mm.round(2))

df['Education_Encoded'] = LabelEncoder().fit_transform(df['Education'])  # text to numbers 0, 1, 2 ...
print("Label Encoded Education:")
print(df[['Education', 'Education_Encoded']])

df_onehot = pd.get_dummies(df['City'], prefix='City').astype(int)  # one 0/1 column per city
print("One-Hot Encoded City:")
print(df_onehot)

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
axes[0].hist(df['Salary'], color='red', edgecolor='black', bins=5)
axes[0].set_title('Original Salary'); axes[0].grid(True)
axes[1].hist(df_std['Salary_std'], color='green', edgecolor='black', bins=5)
axes[1].set_title('StandardScaled'); axes[1].grid(True)
axes[2].hist(df_mm['Salary_mm'], color='blue', edgecolor='black', bins=5)
axes[2].set_title('MinMaxScaled'); axes[2].grid(True)
plt.tight_layout(); plt.show()
```
