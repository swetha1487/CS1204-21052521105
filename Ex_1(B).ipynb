```python
import pandas as pd
```

```python
s = pd.Series([10, 20, 30, 40, 50])              
print("Series:")
print(s)

data = {'Name': ['Arun', 'Bala', 'Chitra', 'Divya', 'Esha'],
        'Age' : [20, 21, 22, 20, 23],
        'Marks': [85, 72, 90, 78, 88]}
df = pd.DataFrame(data)                          
print("DataFrame:")
print(df)

print("Describe:")
print(df.describe())                            
print("Names:", df['Name'].tolist())  
print("First 2 rows:")
print(df.head(2))                                
print("Sorted by Marks:")
print(df.sort_values('Marks', ascending=True))  
print("Mean marks by Age:")
print(df.groupby('Age')['Marks'].mean())   
print("Highest Marks:", df['Marks'].max())     
```
