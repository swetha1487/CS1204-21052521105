```python
import numpy as np
import os

a = np.array([1, 2, 3, 4])
b = np.array((5, 6, 7, 8))                                  
print("From list :", a)
print("From tuple:", b)

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("b / a =", b / a)
m = np.array([[1, 2, 3], [4, 5, 6]])
print("Transpose:")
print(m.T)                                                

c = np.array([10, 20, 30, 40])
f = c * 9 / 5 + 32                                          
print("Fahrenheit:", f)
print("Centigrade:", (f - 32) * 5 / 9)

z = np.array([1 + 2j, 3 - 4j])
print("Real:", z.real, " Imaginary:", z.imag)               

d = np.array([1, 2, 3], ndmin=3)                           
print("Dimensions:", d.ndim)

t = np.array([True, False, True, False])
print("Sorted boolean:", np.sort(t))                        

os.makedirs('data',exist_ok=True)
np.savetxt('data/numbers.csv', m, delimiter=',', fmt='%d')  
print("Saved to data/numbers.csv")
```
